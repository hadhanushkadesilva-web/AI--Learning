```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")        # nano model (~6MB, pretrained)
model.train(
    data="coco128.yaml",          # ← auto-downloads this dataset
    epochs=10,
    imgsz=640
)
```

```python
results = model("https://ultralytics.com/images/bus.jpg")
results[0].show()
```
